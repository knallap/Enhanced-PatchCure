```python
import matplotlib.pyplot as plt

# Data
k_values = list(range(13))
clean_acc = [0.8324, 0.832, 0.832, 0.829, 0.8298, 0.8294, 0.8274, 0.8256, 0.8152, 0.808, 0.8052, 0.7994, 0.7958]
certified_acc = [0.6988, 0.6978, 0.696, 0.6962, 0.6956, 0.6928, 0.6886, 0.6844, 0.6814, 0.669, 0.6632, 0.6646, 0.6424]
throughput = [282.2, 287.57, 288.27, 290.79, 291.08, 293.616, 293.23, 295.323, 298.43,303.41, 303.97, 307.26, 309.46]

# Convert accuracy to percentage
clean_acc = [x * 100 for x in clean_acc]
certified_acc = [x * 100 for x in certified_acc]

# Plot
fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.set_xlabel('Split Layer (k)')
ax1.set_ylabel('Accuracy (%)', color='tab:blue')
ax1.plot(k_values, clean_acc, label='Clean Accuracy', marker='o', color='tab:blue')
ax1.plot(k_values, certified_acc, label='Certified Robust Accuracy', marker='o', color='tab:orange')
ax1.tick_params(axis='y', labelcolor='tab:blue')
ax1.set_ylim(0, 90)

ax2 = ax1.twinx()
ax2.set_ylabel('Throughput (img/s)', color='tab:green')
ax2.plot(k_values, throughput, label='Throughput', marker='o', color='tab:green')
ax2.tick_params(axis='y', labelcolor='tab:green')
ax2.set_ylim(280, 310)

# Legends
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines + lines2, labels + labels2, loc='center right')

plt.title('Enhanced PatchCURE Performance Across Split Layers (k)')
plt.tight_layout()
plt.show()
```
